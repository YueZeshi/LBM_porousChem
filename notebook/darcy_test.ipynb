```python

import numpy as np
import matplotlib.pyplot as plt
```

可用来检验LBM算法在多孔介质中的行为

```python
y = np.linspace(-0.5,0.5,100)
for Da in np.logspace(-4,0,5):
    # k = 0.001
    v_x = (1-np.cosh(np.sqrt(1/Da)*y)/np.cosh(0.5*np.sqrt(1/Da)))/(1-1/np.cosh(0.5*np.sqrt(1/Da)))
    plt.plot(y,v_x,label=f'Da={Da:.4f}')
plt.legend()
plt.show()
```

```python
y = np.linspace(-0.5,0.5,100)
for k in np.logspace(-4,0,5):
    # k = 0.001
    v_x = (1-np.cosh(np.sqrt(1/k)*y)/np.cosh(0.5*np.sqrt(1/k)))/(1-1/np.cosh(0.5*np.sqrt(1/k)))
    plt.plot(y,v_x,label=f'k={k:.4f}')
plt.legend()
plt.show()
```
